## 1. Dataset Creation and Version Management
We will generate 250 synthetic records representing normal conditions (Version 1), and another 250 records representing summer conditions (Version 2) with higher temperatures and delays.

```python
import numpy as np
import pandas as pd

def generate_milk_data(seed, ambient_temp_mean, transport_delay_mean):
    np.random.seed(seed)
    n_samples = 250

    # Generate features
    milk_quantity = np.random.uniform(50, 1000, n_samples) # litres
    collection_temp = np.random.uniform(3.0, 8.0, n_samples) # °C
    transportation_duration = np.random.exponential(scale=transport_delay_mean, size=n_samples) # hours
    cooling_delay = np.random.uniform(0.1, 3.0, n_samples) # hours
    milk_fat_percentage = np.random.uniform(3.0, 5.5, n_samples) # %
    measured_acidity = np.random.uniform(6.4, 6.9, n_samples) # pH
    container_type = np.random.choice(['Plastic Drum', 'Stainless Steel Tank', 'Aluminum Can'], size=n_samples)
    ambient_temperature = np.random.normal(loc=ambient_temp_mean, scale=3, size=n_samples) # °C

    df = pd.DataFrame({
        'milk_quantity': milk_quantity,
        'collection_temp': collection_temp,
        'transportation_duration': transportation_duration,
        'cooling_delay': cooling_delay,
        'milk_fat_percentage': milk_fat_percentage,
        'measured_acidity': measured_acidity,
        'container_type': container_type,
        'ambient_temperature': ambient_temperature
    })

    # Plausible quality relationships for batch rejection
    # High acidity (lower pH < 6.5), high temperatures, and delays increase rejection probability
    rejection_score = (
        (6.7 - df['measured_acidity']) * 3.0 +
        (df['collection_temp'] - 4.0) * 0.5 +
        (df['transportation_duration'] * 0.4) +
        (df['cooling_delay'] * 0.8) +
        ((df['ambient_temperature'] - 20) * 0.1)
    )

    # Map to Yes/No
    df['batch_rejected'] = np.where(rejection_score > 1.5, 'Yes', 'No')
    return df

# Version 1: Normal conditions
df_v1 = generate_milk_data(seed=42, ambient_temp_mean=18, transport_delay_mean=1.5)
df_v1.to_csv('milk_data_v1.csv', index=False)
print("Version 1 Data (Normal Conditions) Created. Rejection Distribution:")
print(df_v1['batch_rejected'].value_counts())

# Version 2: Summer conditions (Higher temperatures and longer delays)
df_v2 = generate_milk_data(seed=101, ambient_temp_mean=32, transport_delay_mean=3.0)
df_v2.to_csv('milk_data_v2.csv', index=False)
print("\nVersion 2 Data (Summer Conditions) Created. Rejection Distribution:")
print(df_v2['batch_rejected'].value_counts())
```

### Explanation of Cell 2: Dataset Creation
Here, we generate simulated data for 250 milk batches to mimic a real-world scenario:
1. **Features**: We randomly generate values for milk quantity, collection temperature, transit time, cooling delays, milk fat percentage, acidity, container type, and ambient temperature.
2. **Rules**: We create a formula where higher transit times, high temperatures, and lower acidity (pH) increase the chances of the milk spoiling.
3. **Version 1 (Normal)**: Represents normal conditions (cooler weather, shorter transit times).
4. **Version 2 (Summer)**: Represents harsher summer conditions with higher temperatures and longer delays, which naturally results in almost all batches being rejected.

## 2. Data Pipeline & Feature Store (Engineering & Preprocessing)
We will create a helper function that constructs our engineered features:
* `temperature_exposure_index`: `collection_temp` * `ambient_temperature`
* `transport_delay_risk`: `transportation_duration` * `cooling_delay`

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

def run_pipeline(df, is_train=True, preprocessor=None):
    # Feature Engineering
    df_feat = df.copy()
    df_feat['temperature_exposure_index'] = df_feat['collection_temp'] * df_feat['ambient_temperature']
    df_feat['transport_delay_risk'] = df_feat['transportation_duration'] * df_feat['cooling_delay']

    # Define column lists
    num_cols = ['milk_quantity', 'collection_temp', 'transportation_duration', 'cooling_delay',
                'milk_fat_percentage', 'measured_acidity', 'ambient_temperature',
                'temperature_exposure_index', 'transport_delay_risk']
    cat_cols = ['container_type']

    # Align target variable
    y = df_feat['batch_rejected'].map({'Yes': 1, 'No': 0}).values
    X_raw = df_feat[num_cols + cat_cols]

    if is_train:
        # Set up transformation pipeline
        preprocessor = ColumnTransformer(
            transformers=[
                ('num', StandardScaler(), num_cols),
                ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols)
            ]
        )
        X_processed = preprocessor.fit_transform(X_raw)
        return X_processed, y, preprocessor
    else:
        X_processed = preprocessor.transform(X_raw)
        return X_processed, y

# Prepare features from both versions to show pipeline reuse
X_v1, y_v1, preprocessor_v1 = run_pipeline(df_v1, is_train=True)
X_v2, y_v2 = run_pipeline(df_v2, is_train=False, preprocessor=preprocessor_v1)

print(f"Engineered feature shape for Version 1: {X_v1.shape}")
print(f"Engineered feature shape for Version 2: {X_v2.shape}")
```

### Explanation of Cell 3: Data Preprocessing Pipeline
Raw data cannot be fed directly into machine learning models. In this cell, we prepare the data:
1. **Feature Engineering**: We combine features to create new indicators:
   * `temperature_exposure_index`: Collection temperature multiplied by ambient temperature (overall heat risk).
   * `transport_delay_risk`: Transportation duration multiplied by cooling delay (time risk).
2. **Scaling & Encoding**: We scale numerical values to have a standard range (so no single feature dominates the model) and convert categorical columns (like container type) into numbers.

## 3. Model Training & Registry
We train Logistic Regression and Random Forest models, evaluate their metrics, and track our training with MLflow.

```python
import mlflow
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, precision_recall_fscore_support
from sklearn.model_selection import train_test_split

# Split Version 1 for validation
X_train, X_val, y_train, y_val = train_test_split(X_v1, y_v1, test_size=0.2, random_state=42)

mlflow.set_experiment("Milk_Batch_Inspection")

models = {
    'Logistic_Regression': LogisticRegression(max_iter=1000),
    'Random_Forest': RandomForestClassifier(random_state=42, n_estimators=100)
}

best_model_name = None
best_f1 = 0.0
best_model_obj = None

for name, model in models.items():
    with mlflow.start_run(run_name=name):
        model.fit(X_train, y_train)
        preds = model.predict(X_val)

        precision, recall, f1, _ = precision_recall_fscore_support(y_val, preds, average='binary')

        # Log params and metrics
        mlflow.log_params(model.get_params())
        mlflow.log_metric("precision", precision)
        mlflow.log_metric("recall", recall)
        mlflow.log_metric("f1_score", f1)

        print(f"=== {name} Performance ===")
        print(classification_report(y_val, preds))

        if f1 > best_f1:
            best_f1 = f1
            best_model_name = name
            best_model_obj = model

# Register the best model to MLflow Run
with mlflow.start_run(run_name="Best_Registered_Model") as run:
    mlflow.sklearn.log_model(best_model_obj, artifact_path="model", registered_model_name="MilkRejectionClassifier")
    print(f"Registered {best_model_name} as 'MilkRejectionClassifier' in MLflow Registry.")
```

### Explanation of Cell 4: Model Training and Selection
In this cell, we train two different algorithms to see which one performs better at predicting milk rejection:
1. **Logistic Regression**: A simple statistical model.
2. **Random Forest**: An ensemble model made of multiple decision trees.
3. **Validation**: We split our data into training and validation sets to test the models on unseen data.
4. **MLflow Tracking & Registry**: We record performance metrics (precision, recall, F1-score) for both models. Since Logistic Regression performed exceptionally well, we automatically register it in MLflow's Model Registry as the champion model.

## 4. CI/CD and Model Serving
First, we will build a FastAPI app and define input validation rules using Pydantic. Then we simulate GitHub Actions runner tests evaluating invalid input conditions.

```python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field, field_validator
from typing import Literal
import nest_asyncio
import uvicorn
import httpx
import threading

# Define the input payload structure
class MilkBatchInput(BaseModel):
    milk_quantity: float = Field(..., description="Quantity in litres")
    collection_temp: float = Field(..., description="Collection temperature in °C")
    transportation_duration: float = Field(..., description="Transportation duration in hours")
    cooling_delay: float = Field(..., description="Cooling delay in hours")
    milk_fat_percentage: float = Field(..., description="Milk fat percentage")
    measured_acidity: float = Field(..., description="Acidity indicator")
    container_type: Literal['Plastic Drum', 'Stainless Steel Tank', 'Aluminum Can']
    ambient_temperature: float = Field(..., description="Ambient temperature in °C")

    @field_validator('milk_quantity')
    @classmethod
    def validate_quantity(cls, v):
        if v < 0:
            raise ValueError('Milk quantity cannot be negative.')
        return v

    @field_validator('collection_temp', 'ambient_temperature')
    @classmethod
    def validate_temperatures(cls, v):
        if v < -273.15 or v > 100.0:
            raise ValueError('Temperature is physically impossible.')
        return v

app = FastAPI(title="Milk Quality Inspection Service")

@app.post("/predict")
def predict(data: MilkBatchInput):
    # Format input payload into Pandas DataFrame
    input_dict = data.model_dump()
    input_df = pd.DataFrame([input_dict])

    # Process data using our reuseable pipeline transformer
    try:
        X_serv, _ = run_pipeline(input_df, is_train=False, preprocessor=preprocessor_v1)
        prediction = best_model_obj.predict(X_serv)[0]
        result = "Rejection Risk" if prediction == 1 else "Likely Accepted"
        return {"prediction": result}
    except Exception as e:
        raise HTTPException(status_code=500, detail=str(e))

# Unit Tests for CI/CD workflow simulation
def run_unit_tests():
    print("--- Running CI/CD Validation Tests ---")

    # Test case 1: Negative Quantity validation
    try:
        MilkBatchInput(
            milk_quantity=-10.0,
            collection_temp=4.5,
            transportation_duration=1.2,
            cooling_delay=0.5,
            milk_fat_percentage=4.0,
            measured_acidity=6.6,
            container_type='Stainless Steel Tank',
            ambient_temperature=20.0
        )
    except ValueError as e:
        print("Test Passed: Successfully caught invalid negative milk quantity!")

    # Test case 2: Impossible Temperature validation
    try:
        MilkBatchInput(
            milk_quantity=100.0,
            collection_temp=120.0,
            transportation_duration=1.2,
            cooling_delay=0.5,
            milk_fat_percentage=4.0,
            measured_acidity=6.6,
            container_type='Stainless Steel Tank',
            ambient_temperature=20.0
        )
    except ValueError as e:
        print("Test Passed: Successfully caught impossible collection temperature!")

run_unit_tests()
```

### Explanation of Cell 5: FastAPI App & CI/CD Validation
This cell prepares our model to be served as an API and simulates automated checks (CI/CD):
1. **Data Validation (Pydantic)**: We define a schema for incoming requests. If a request has negative milk quantities or impossible temperatures (like 120°C), it automatically raises an error.
2. **FastAPI Server**: We create a `/predict` endpoint that receives batch details, processes them using our pipeline, and outputs "Rejection Risk" or "Likely Accepted".
3. **Unit Tests**: We run automated checks to confirm that our validator correctly catches bad inputs before they can reach the machine learning model.

### 5. Start serving application & verify the endpoint
We launch the server asynchronously in the background and request sample payloads using `httpx`.

```python
# Apply nest_asyncio to run uvicorn loop inside Google Colab
nest_asyncio.apply()

def start_server():
    uvicorn.run(app, host="127.0.0.1", port=8000, log_level="warning")

# Spin up server in background thread
server_thread = threading.Thread(target=start_server, daemon=True)
server_thread.start()

# Give the server a moment to spin up
import time
time.sleep(2)

# Create a client to test valid request scenarios
with httpx.Client(base_url="http://127.0.0.1:8000") as client:
    # Normal sample expected to be Accepted
    sample_good = {
        "milk_quantity": 250.0,
        "collection_temp": 3.5,
        "transportation_duration": 0.5,
        "cooling_delay": 0.2,
        "milk_fat_percentage": 4.1,
        "measured_acidity": 6.8,
        "container_type": "Stainless Steel Tank",
        "ambient_temperature": 15.0
    }

    # Critical Summer/Delay scenario
    sample_bad = {
        "milk_quantity": 250.0,
        "collection_temp": 9.5,
        "transportation_duration": 8.5,
        "cooling_delay": 3.2,
        "milk_fat_percentage": 3.5,
        "measured_acidity": 6.1,
        "container_type": "Plastic Drum",
        "ambient_temperature": 36.0
    }

    res_good = client.post("/predict", json=sample_good)
    res_bad = client.post("/predict", json=sample_bad)

    print("\n--- Verification on Live Server ---")
    print("Normal Batch Response:", res_good.json())
    print("High Risk Batch Response:", res_bad.json())
```
